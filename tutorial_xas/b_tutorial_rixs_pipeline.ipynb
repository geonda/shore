```python
import shore
```

```python
photon1=shore.Photon(interaction='dipole', polarization=[0,0,1],energy=dict(element='C', edge='K'))
photon2=shore.Photon(interaction='dipole', polarization=[0,1,0],energy=dict(element='C', edge='K'))
```

```python
beam=shore.Light(photons=[photon1, photon2])
```

```python
c=shore.AtomicStructure('data/structures/c.cif')
material=shore.Matter(structure=c, ground_state=dict(bands=24, kmesh=-4, ecut=-1, nspin=1, starting_magnetization={'1':0}), 
                        screening=dict(bands=80, kmesh=-2, diemac=4.7),
                        bse=dict(bands=30, kmesh=-4),
                        rixs=dict(photon_in=1, photon_out=2, core_broad=0.4, val_broad=0.4, range='10 30 2'),
                        prefix='srun -n 16')
```

```python
ocean=shore.Input(name='rixs', matter=material, light=beam)
```

```python
server=shore.RemoteServerManager(load='../test.server.pkl')
```

```python
rixs=shore.Calculation(input=ocean,server=server)
```
