```python
import shore
```

```python
photon1=shore.Photon(interaction='dipole', polarization=[0,0,1],energy=dict(element='C', edge='K'))
photon2=shore.Photon(interaction='dipole', polarization=[0,1,0],energy=dict(element='C', edge='K'))
```

```python
beam=shore.Light(photons=[photon1, photon2])
```

```python
diamond=shore.AtomicStructure('data/structures/c.cif')
material=shore.Matter(structure=diamond, ground_state=dict(bands=24, kmesh=-4, ecut=-1, nspin=1, starting_magnetization={'1':0}), 
                        screening=dict(bands=80, kmesh=-2, diemac=4.7),
                        bse=dict(bands=30, kmesh=-4),
                        xas=dict(broad=0.4, range="10000 -10 40"),
                        prefix='srun -n 16')
```

```python
ocean=shore.Input(name='xas', matter=material, light=beam)
```

```python
hpc=shore.RemoteServerManager(load='../test.server.pkl')
```

```python
xas=shore.Calculation(input=ocean,server=hpc)
```

```python
xas.run()
```

```python
xas.get_status()
```

```python
results=xas.sync()
```

```python
from plotly import graph_objects as go
fig=go.Figure()
fig=results.plot_xas(fig=fig)
fig.update_xaxes(range=[0,40])
fig.show()
```
